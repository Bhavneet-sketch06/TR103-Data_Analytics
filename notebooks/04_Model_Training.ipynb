# Business Insights Dashboard with AI Sales Prediction

## Notebook 04 - Model Training

### Objectives
- Import required libraries
- Load the cleaned dataset
- Perform feature engineering
- Train the Linear Regression model
- Evaluate the model
- Save the trained model using Joblib

```python
# Data Manipulation
import pandas as pd
import numpy as np

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Model Evaluation
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Save Model
import joblib
```

```python
df = pd.read_csv("../data/Sample - Superstore.csv", encoding="latin1")

df.head()
```

```python
# Convert to datetime

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Year

df["Year"] = df["Order Date"].dt.year

# Month

df["Month"] = df["Order Date"].dt.month

# Month Name

df["Month Name"] = df["Order Date"].dt.month_name()

# Shipping Days

df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df.columns
```

```python
features = [
    "Category",
    "Sub-Category",
    "Ship Mode",
    "Region",
    "State",
    "Segment",
    "Quantity",
    "Discount",
    "Profit",
    "Year",
    "Month",
    "Shipping Days"
]
```

```python
X = df[features]

y = df["Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
categorical_features = [
    "Category",
    "Sub-Category",
    "Region",
    "Segment",
    "Ship Mode",
    "State"
]

numerical_features = [
    "Quantity",
    "Discount",
    "Profit",
    "Year",
    "Month",
    "Shipping Days"
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numerical_features
        )
    ]
)
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
```

```python
model.fit(X_train, y_train)
```

# Step 12: Model Evaluation

The trained model is evaluated on the test dataset.

Predictions are made on unseen data, and different evaluation metrics are calculated to measure the model's performance.

```python
# Make predictions on the test dataset

y_pred = model.predict(X_test)
```

```python
comparison = pd.DataFrame({
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

comparison.head(10)
```

## Mean Absolute Error (MAE)

MAE calculates the average absolute difference between the actual and predicted values.

A lower MAE indicates better model performance.

```python
mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
```

## Mean Squared Error (MSE)

MSE calculates the average squared difference between actual and predicted values.

Larger errors receive a higher penalty because the differences are squared.

```python
mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error (MSE):", round(mse, 2))
```

```python
rmse = np.sqrt(mse)

print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

## R² Score

The R² Score measures how well the model explains the variation in the target variable.

The closer the value is to 1, the better the model fits the data.

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", round(r2, 4))
```

```python
print("=" * 50)
print("MODEL EVALUATION")
print("=" * 50)

print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
# Create Random Forest Pipeline

rf_model = Pipeline(

    steps=[

        ("preprocessor", preprocessor),

        ("regressor",
         RandomForestRegressor(
             n_estimators=100,
             random_state=42
         ))
    ]
)
```

```python
rf_model.fit(X_train, y_train)
```

# Random Forest Model Evaluation

```python
# Make predictions on the test data

rf_predictions = rf_model.predict(X_test)
```

```python
rf_comparison = pd.DataFrame({

    "Actual Sales": y_test.values,

    "Predicted Sales": rf_predictions

})

rf_comparison.head(10)
```

```python
rf_mae = mean_absolute_error(y_test, rf_predictions)

print("Random Forest MAE :", round(rf_mae,2))
```

```python
rf_mse = mean_squared_error(y_test, rf_predictions)

print("Random Forest MSE :", round(rf_mse,2))
```

```python
rf_rmse = np.sqrt(rf_mse)

print("Random Forest RMSE :", round(rf_rmse,2))
```

```python
rf_r2 = r2_score(y_test, rf_predictions)

print("Random Forest R² :", round(rf_r2,4))
```

```python
print("="*55)
print(" RANDOM FOREST MODEL EVALUATION ")
print("="*55)

print(f"MAE  : {rf_mae:.2f}")
print(f"MSE  : {rf_mse:.2f}")
print(f"RMSE : {rf_rmse:.2f}")
print(f"R²   : {rf_r2:.4f}")
```

# Model Comparison

```python
comparison_df = pd.DataFrame({

    "Metric": ["MAE","MSE","RMSE","R² Score"],

    "Linear Regression":[
        mae,
        mse,
        rmse,
        r2
    ],

    "Random Forest":[
        rf_mae,
        rf_mse,
        rf_rmse,
        rf_r2
    ]

})

comparison_df
```

# Model Comparison & Selection

Two regression algorithms were trained and evaluated.

1. Linear Regression
2. Random Forest Regressor

The evaluation metrics show that Random Forest performs significantly better than Linear Regression.

Therefore, Random Forest is selected as the final model for deployment in the Streamlit application.
