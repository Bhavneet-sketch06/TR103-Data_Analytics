```python
import pandas as pd
df = pd.read_csv("../data/Sample - Superstore.csv", encoding="latin1")
df_clean = df.copy()
```

```python
df_clean.head()
```

```python
df_clean.dtypes
```

```python
df_clean["Order Date"] = pd.to_datetime(df_clean["Order Date"])
df_clean["Ship Date"] = pd.to_datetime(df_clean["Ship Date"])
df_clean.dtypes
```

```python
df_clean["Year"] = df_clean["Order Date"].dt.year 
df_clean["Month"] = df_clean["Order Date"].dt.month
df_clean[["Order Date", "Year"]].head()
df_clean[["Order Date" , "Month"]].head()
```

```python
df_clean["Month Name"] = df_clean["Order Date"].dt.month_name()
df_clean[["Order Date" , "Month Name"]].head()
```

```python
df_clean["Day"] = df_clean["Order Date"].dt.day
df_clean["Day Name"] = df_clean["Order Date"].dt.day_name()
df_clean[["Order Date" , "Day"]].head()
df_clean[["Order Date" , "Day Name"]].head()
```

```python
df_clean.head()
```

```python
df_clean.info()
```

# Exploratory Data Analysis (EDA)

In this section, we analyze the business data to identify sales trends, customer behavior, product performance, and regional performance.

```python
total_sales = df_clean["Sales"].sum()
print(f"Total Sales : ${total_sales:,.2f}")
```

```python
total_profit = df_clean["Profit"].sum()
print(f"Total Profit : ${total_profit:,.2f}")
```

```python
total_orders = df_clean["Order ID"].nunique()
print("Total Orders :" , total_orders)
```

```python
total_customers = df_clean["Customer ID"].nunique()
print("Total Customers :",total_customers)
```

```python
total_products = df_clean["Product Name"].nunique()
print("Total Products:", total_products)
```

```python
average_sales = df_clean["Sales"].mean()
print(f"Average Sales: ${average_sales:.2f}")
```

```python
average_profit = df_clean["Profit"].mean()
print(f"Average Profit: ${average_profit:.2f}")
```

```python
largest_sale = df_clean["Sales"].max()
print(f"Largest Sale: ${largest_sale:,.2f}")
```

```python
smallest_sale = df_clean["Sales"].min()
print(f"Smallest Sale: ${smallest_sale:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
```

```python
category_sales = df.groupby("Category").agg({"Sales" : "sum"})
print(category_sales)
```

```python
import matplotlib.pyplot as plt

category_sales = df_clean.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))

bars = plt.bar(category_sales.index,
               category_sales.values)

plt.title("Total Sales by Category",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Category", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

# Display values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"${height:,.0f}",
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.tight_layout()

plt.show()
```

### Observation

Highest number of sales is through the Technology Category.
Lowest number of sales is in case of Office Supplies Category.
while Furniture also performs strongly.
